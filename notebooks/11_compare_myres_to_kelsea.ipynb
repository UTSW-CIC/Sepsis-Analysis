```python
df_agg_vent = df_agg_vent.sort(by=['EncounterEpicCsn', 'Event_DateTime'])
df_organdysfunction = df_organdysfunction.sort(by=['EncounterEpicCsn', 'Event_DateTime'])
df_septicshock = df_septicshock.sort(by=['EncounterEpicCsn', 'Event_DateTime'])

df_agg_vent.select(
    "EncounterEpicCsn",
	"Event_DateTime",
).equals(
	df_organdysfunction.select(
		"EncounterEpicCsn",
		"Event_DateTime",
	)
), df_organdysfunction.select(
    "EncounterEpicCsn",
	"Event_DateTime",
).equals(
	df_septicshock.select(
		"EncounterEpicCsn",
		"Event_DateTime",
	)
)
```

```python
df_mo = df_agg_vent
cols = set(df_mo.columns)
df_mo = pl.concat([df_mo, df_organdysfunction.select(set(df_organdysfunction.columns)-set(df_mo.columns))], how='horizontal')
df_mo = pl.concat([df_mo, df_septicshock.select(set(df_septicshock.columns)-set(df_mo.columns))], how='horizontal')

df_mo
```

```python
vitals = [
    'temp',
    'pulse',
    'respirations',
    'wbc',
    'map',
    'sbp'
]

labs = [
    'egfr',
	'creatinine',
	'lactate',
	'Platelet',
	'bilirubin',
	'inr',
    # 'aptt',
    'urine',
]
```

```python
def get_column_pairs(event_name, df_cols):
	return [
		c  for c in df_cols if event_name.lower() in c.lower() and ('Most' in c or 'last' in c)
	]

get_column_pairs(labs[3], df_all.columns)
```

```python
for l in vitals:
	col_pairs = get_column_pairs(l, df_all.columns)
	if len(col_pairs) != 2:
		print(f'{l} does not have 2 matching columns. {col_pairs} found ...')
		continue
	n_rows = df_all.filter(
		pl.col("EncounterEpicCsn").is_not_null()&pl.col("EncounterEpicCsn_right").is_not_null()
	).shape[0]

	print(l)
	print(df_all.filter(
		pl.col("EncounterEpicCsn").is_not_null()&pl.col("EncounterEpicCsn_right").is_not_null()
	).select(
		["EncounterEpicCsn",
		"EncounterEpicCsn_right",
		"Event_DateTime",
		"Event_DateTime_right"]+
		col_pairs
	).with_columns(
		(pl.col(col_pairs[0])==pl.col(col_pairs[1])).mean().alias('match_rate'),
		( (pl.col(col_pairs[0])!=pl.col(col_pairs[1])).sum()/n_rows ).alias('num_mismatch_match'),

		pl.corr(col_pairs[0], col_pairs[1]).alias('correlation'),
		(pl.col(col_pairs[0]) - pl.col(col_pairs[1])).abs().mean().alias('difference')
	).select("match_rate", "num_mismatch_match", "correlation", "difference")[0])
	print("===============================================")
```
